# Analisis del catalogo de libros

Exploracion con pandas, grafico y conclusiones.

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_json("datos.json")
df
```

```python
df.info()
```

```python
df.describe()
```

```python
por_genero = df.groupby("genero")["titulo"].count().sort_values(ascending=False)
por_genero
```

```python
por_genero.plot(kind="bar", color="steelblue", figsize=(8, 5))
plt.title("Cantidad de libros por genero")
plt.xlabel("Genero")
plt.ylabel("Cantidad")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```

```python
df["calificacion"].plot(kind="hist", bins=10, color="darkorange", figsize=(8, 5))
plt.title("Distribucion de calificaciones")
plt.xlabel("Calificacion")
plt.show()
```

## Conclusiones

- El genero predominante es Novela.
- La calificacion promedio es alta (cerca de 4.5), indicando un catalogo bien valorado.
- El valor total del stock se calcula como la suma de precio por stock.
